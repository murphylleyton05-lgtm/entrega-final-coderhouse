## 5. Reproducir el video acá mismo (con audio) y descargarlo

```python
import glob
from base64 import b64encode
from IPython.display import HTML, display
vid = sorted(glob.glob('output/*/video.mp4'))[-1]
data = b64encode(open(vid, 'rb').read()).decode()
display(HTML(f'<video width=340 controls><source src="data:video/mp4;base64,{data}" type="video/mp4"></video>'))
print('Archivo:', vid)
```
